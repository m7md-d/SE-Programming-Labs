Write a program that:

- Creates an integer, string, list, and dictionary.
- Uses the `+` operator where supported.
- Uses the `*` operator where supported.
- Uses `len()` where supported.
- Uses the `in` operator where supported.
- Uses a `for` loop  where supported.

````python
counter = 12
name = "no"
one = [1, 2, 3, 4, 5]
two = {"name": 1, "jo": 2}

for n, b in two.items():
    x = (len(one) + counter) * name
    print(n,": ",b)
print(x)
````
