## Practice
Create a class named `MobilePhone` with:
- A class variable named `category` and assign it the value "Electronics". 
- Three instance variables: `brand`, `model`, `price`.
- An instance method `display_details()` that displays the phone's details.
- A class method `change_category()` that accepts a new category value and updates the class variable.
- Create two objects with different values.
- Call the `display_details()` method for both objects.
- Use the `change_category()` method to change the category from "Electronics" to "Smart Devices" and verify that both objects reflect the update.

```python
class MobilePhone:
    category = "Electronics"
    def __init__(s, brand, model, price):
        s.brand = brand
        s.model = model
        s.price = price
    def display_details(s):
        print(f"The phone brand: {s.brand}")
        print(f"The phone model: {s.model}")
        print(f"The phone price: {s.price}")
        print(f"The phone price: {s.category}")
    def change_category(s, new_category):
        s.category = new_category
iphone = MobilePhone("apple", "13 pro", 999)
samsung = MobilePhone("samsung", "25 edge", 600)

iphone.display_details()
samsung.change_category("Smart Devices")
iphone.display_details()
```
